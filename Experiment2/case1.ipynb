```python
import cv2
import numpy as np

# Read grayscale image
img = cv2.imread("../exp01/images.jpg", 0)

# Check image
if img is None:
    print("Image not found!")
    exit()

# Calculate mean threshold
threshold = np.mean(img)

print("Threshold =", threshold)

# Convert to binary image
binary = np.where(img > threshold, 255, 0).astype(np.uint8)

# Display images
cv2.imshow("Original Grayscale", img)
cv2.imshow("Binary Image", binary)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# que 2 : 
# Case 1 : Convert an RGB Image to Grayscale using Mean Average
import cv2
import numpy as np

# Read color image
img = cv2.imread("../exp01/images.jpg", 1)

# Check image
if img is None:
    print("Image not found!")
    exit()

# Split into B, G, R channels
B, G, R = cv2.split(img)

# Convert to grayscale using mean
gray = ((R + G + B) / 3).astype(np.uint8)

# Display images
cv2.imshow("Original Color Image", img)
cv2.imshow("Grayscale Image", gray)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Convert an RGB (color) image into a grayscale image.

# Case 2: Input the weightages of R, G, and B from the user. 
# Each weight should be between 0 and 1, and the sum of all three weights must be 1.
import cv2
import numpy as np

# Read color image
img = cv2.imread("../exp01/images.jpg", 1)

# Check image
if img is None:
    print("Image not found!")
    exit()

# Input weights
wr = float(input("Enter weight for Red: "))
wg = float(input("Enter weight for Green: "))
wb = float(input("Enter weight for Blue: "))

# Check if weights sum to 1
if abs((wr + wg + wb) - 1) > 0.0001:
    print("Error! Sum of weights must be 1.")
    exit()

# Split image into B, G, R channels
B, G, R = cv2.split(img)

# Convert to grayscale
gray = (R * wr + G * wg + B * wb).astype(np.uint8)

# Display images
cv2.imshow("Original Color Image", img)
cv2.imshow("Grayscale Image", gray)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
