```python
import cv2
import numpy as np

img = cv2.imread("images.jpg")

print(img)
```

```python
import matplotlib.pyplot as plt
# Convert BGR to RGB
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img)
plt.axis("off")
plt.show()
```

```python
cv2.imwrite("image.png", img)
cv2.imwrite("image.tiff", img)
cv2.imwrite("image.bmp", img)
cv2.imwrite("image.jpeg", img)
cv2.imwrite("image.webp", img)

print("Image converted into PNG, TIFF, BMP, JPEG, and WEBP formats successfully.")
```

```python
import cv2

# Read video
video = cv2.VideoCapture("video.mp4")

while True:
    ret, frame = video.read()

    if ret == False:
        break

    cv2.imshow("Video", frame)

    if cv2.waitKey(25) == ord('q'):
        break

video.release()
cv2.destroyAllWindows()
```

```python
# Extract frames from a video.
import os
video = cv2.VideoCapture("video.mp4")
output_folder = "frames"
os.makedirs(output_folder, exist_ok=True)

frame_count = 0

while True:
    ret, frame = video.read()
    if not ret:
        break

    frame_name = os.path.join(output_folder, f"frame_{frame_count}.jpg")
    cv2.imwrite(frame_name, frame)
    frame_count += 1

video.release()
print(f"{frame_count} frames extracted and saved in '{output_folder}' folder.")
```
